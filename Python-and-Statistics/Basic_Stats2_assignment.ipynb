```python
from scipy import stats
import pandas as pd
import numpy as np
```

```python
durability = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
n = len(durability)
sample_mean = np.mean(durability)
degree_of_freedom = len(durability) - 1
sample_stddev = np.std(durability)
std_error = sample_stddev / np.sqrt(n)

print('The mean durability of sample is: ', sample_mean)
print('The standard error of sample is: ', std_error)

```

```python
sample_stddev
```

```python
print(stats.t.interval(0.99, df = degree_of_freedom))
print('stats.t.interval(0.99, df = 14)[0] = ',stats.t.interval(0.99, df = degree_of_freedom)[0])
print('stats.t.interval(0.99, df = 14)[1] = ',stats.t.interval(0.99, df = degree_of_freedom)[1])
```

```python
print('Lower Limit of 99% Confidence Interval',round(sample_mean + (stats.t.interval(0.99, df = degree_of_freedom)[0])*(std_error),3))
print('Upper Limit of 99% Confidence Interval',round(sample_mean + (stats.t.interval(0.99, df = degree_of_freedom)[1])*(std_error),3))
```

```python
avg_durability = stats.t.interval(0.99, 14, loc = sample_mean, scale = std_error)
print('The average durability of printheads is between ',round(avg_durability[0],3),'mil',' and ',round(avg_durability[1],3),'mil  characters')
```

```python
import numpy as np
from scipy.stats import norm

# Sample data
durability = np.array([1.13, 1.55, 1.43, 0.92, 1.25, 1.36, 1.32, 0.85, 1.07, 1.48, 1.20, 1.33, 1.18, 1.22, 1.29])
sigma = 0.2  # Given population standard deviation
confidence_level = 0.99  # 99% Confidence Interval

# Sample statistics
x_bar = np.mean(durability)  # Sample mean
n = len(durability)  # Sample size

# Z-score for 99% confidence level
z_score = norm.ppf(1 - (1 - confidence_level) / 2)

# Margin of Error (ME)
margin_error = z_score * (sigma / np.sqrt(n))

# Confidence Interval
lower_bound = x_bar - margin_error
upper_bound = x_bar + margin_error

print(f"Sample Mean: {x_bar}")
print(f"Z-score: {z_score}")
print(f"Margin of Error: {margin_error}")
print(f"99% Confidence Interval: ({lower_bound}, {upper_bound})")
```
